#### Partie 1 – Series 

##### 1) Création d'une Series contenant 4 températures  

```python
serie_temp4 = pd.Series([12.25, 38.01, 5.8, 14.5])
serie_temp4
```

##### 2) Affichage de l'index de la série

```python
print("L'index de la séerie : ",serie_temp4.index)
```

##### 3) Affichage de la 1ère valeur de la série

```python
print("La 1ére valeur de la série est : ", serie_temp4[0])
```

##### 4) Affichage de la dernière valeur de la série

```python
print("La dernière valeur de la série est : ", serie_temp4.iloc[-1])
```

##### 5) Création de deux façon différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h 

```python
serie_temp_h = pd.Series([21.5, 32, 7.03, 18], index=['12h', '13h', '14h', '15h'])
serie_temp_h
```

##### 6) Affichage la 1ère et la dernière valeur 

```python
# 1e valeur
print("La 1ére valeur de la série est : ", serie_temp_h['12h'])

# dernière valeur
print("La dernière valeur de la série est : ", serie_temp_h.iloc[-1])
```
