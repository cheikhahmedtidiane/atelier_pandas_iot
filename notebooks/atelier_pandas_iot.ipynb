#### Partie 5 – Manipulation des colonnes

##### 1) Ajout de la colonne "temperature_fahrenheit" 

```python
# formule F = C x (9/5) + 32
df['temperature_fahrenheit'] = (df['temperature'] * (9/5) + 32)
df.head(3)
```

##### 2) Ajout de la colonne "niveau_temperature" 
"Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy

```python
import numpy as np

df['niveau_temperature'] = np.where(df['temperature'] > 30, 'Élevée', 'Normale')
df.head(3)
```

##### 3) Renommage de la colonne "humidite" en "humidite_relative" 

```python
df = df.rename(columns={'humidite': 'humidite_relative'})
df.head(3)
```

##### 4) Suppression de la colonne "temperature_fahrenheit" 

```python
df.drop(columns='temperature_fahrenheit', inplace=True)
df.head(3)
```

#### Partie 8 : Analyse

##### 1) Affichage de la consommation moyenne par bâtiment

```python
df.groupby('batiment')['consommation'].mean()
```

##### 2) Affichage du minimum, du maximum, de la moyenne et l’écart-type de la température par bâtiment 

```python
# Calcul avec des noms de colonnes personnalisés
stats_temp_bat = df.groupby('batiment')['temperature'].agg(
    Minimum = 'min',
    Maximum = 'max',
    Moyenne = 'mean',
    Ecart_Type = 'std'
)

print(stats_temp_bat)
```

##### 3) Affichage par bâtiment, le maximum et la moyenne de la température et de la consommation

```python
# Agrégation nommée pour chaque colonne spécifique
bat_tempcons = df.groupby('batiment').agg(
    Temp_Max = ('temperature', 'max'),
    Temp_Moyenne = ('temperature', 'mean'),
    Conso_Max = ('consommation', 'max'),
    Conso_Moyenne = ('consommation', 'mean')
)

print(bat_tempcons)
```

##### 4) Affichage la température moyenne globale

```python
temp_moy_glob = df['temperature'].mean()
print(f"Température moyenne globale : {temp_moy_glob:.2f} °C")
```

##### 5) Affichage de la température maximale 

```python
temp_max = df['temperature'].max()
print(f"Température maximale : {temp_max:.2f} °C")
```

##### 6) Affichage de la consommation totale 

```python
cons_total = df['consommation'].sum()
print(f"Consommation totale : {cons_total:.3f}")
```

##### 7) Quel bâtiment consomme le plus en moyenne ? 

```python
# la consommation moyenne par bâtiment
conso_par_batiment = df.groupby('batiment')['consommation'].mean()

# le bâtiment avec la moyenne la plus élevée
batiment_max = conso_par_batiment.idxmax()
valeur_max = conso_par_batiment.max()

print(f"Le bâtiment qui consomme le plus en moyenne est le {batiment_max} avec {valeur_max:.2f}")
```

##### Partie 9 : Gestion des valeurs manquantes

##### 1) Affichage du nombre de valeurs manquantes par colonne 

```python
df.isna().sum()
```

##### 2) Affichage du nombre total des valeurs manquantes 

```python
print("Nombres total de valeurs manquantes : ",df.isna().sum().sum())
```

##### 3) Affichage du taux de valeurs manquantes 

```python
# taux de valeurs manquantes pour chaque colonne (en %)
taux_manquants = df.isna().mean() * 100

print("Taux de valeurs manquantes par colonnes")
print(taux_manquants.round(2))
```

```python
print("Taux de valeurs manquantes total: ", taux_manquants.sum().round(2),"%")
```

##### 4) Affichage des lignes contenant des valeurs manquantes 

```python
lignes_manquantes = df[df.isna().any(axis=1)]
lignes_manquantes
```

#### 5) Pour la température et l’humidité, remplaçons les valeurs manquantes par les moyennes

```python
# Sélection des colonnes cibles et application de fillna avec leur propre moyenne
colonnes_cibles = ['temperature', 'humidite_relative']
df[colonnes_cibles] = df[colonnes_cibles].fillna(df[colonnes_cibles].mean())
df.isna().sum()
```

##### 6) Pour la consommation, remplacer les valeurs manquantes par la médiane

```python
df['consommation'] = df['consommation'].fillna(df['consommation'].median())
df.isna().sum()
```

##### 7) Pour "etat", remplaçons les valeurs manquantes par "INCONNU" 

```python
df['etat'] = df['etat'].fillna('INCONNU')
df.isna().sum()
```
